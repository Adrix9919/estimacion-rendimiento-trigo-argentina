```python
import pandas as pd

df = pd.read_csv(
    "../data/raw/trigo-serie-1927-2025.csv",
    dtype={"provincia_id": str, "departamento_id": str},
    encoding="latin-1",
)
df.shape
```

```python
print(df.dtypes)
print(df["provincia"].unique())
```

```python
print(df["anio"].min(), df["anio"].max())
print(df["departamento_id"].nunique())
print(df.isna().sum())
```

```python
print(df.duplicated(subset=["anio", "provincia_id", "departamento_id"]).sum())

calc = df["produccion_tm"] * 1000 / df["superficie_cosechada_ha"]
print(((df["rendimiento_kgxha"] - calc).abs() > 0.05 * calc).sum())
```

```python
periodo = df[(df["anio"] >= 2000) & (df["anio"] <= 2025)]
print(len(periodo))
print(len(periodo[periodo["anio"] <= 2019]))
print(len(periodo[periodo["anio"] >= 2020]))
```

```python
resumen = df.groupby("anio").agg(
    filas=("anio", "size"),
    departamentos=("departamento_id", "nunique"),
    sin_rendimiento=("rendimiento_kgxha", lambda s: s.isna().sum()),
)
print(resumen.to_string())
```
